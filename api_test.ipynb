# RMV API

## Setup

```python
import os
import requests
from datetime import datetime
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("api_key")
```

```python
domain = "www.rmv.de/hapi"

headers = {
    "Authorization": f"Bearer {api_key}",
    "accept": "application/json",
}
```

## Location search

```python
service = "location.name"
url = f"https://{domain}/{service}"

params = {    
    "input": "Mainz Uni",
}

x = requests.get(url, headers=headers, params=params)
locations = x.json()["stopLocationOrCoordLocation"]
```

```python
locations[0].keys()
```

```python
for location in locations:
    name = location.get("StopLocation", {}).get("name", "N/A")
    id = location.get("StopLocation", {}).get("id", "N/A")
    print(name, id)
```

```python
service = "location.name"
url = f"https://{domain}/{service}"

params = {    
    "input": "Höfchen Mainz",
}

x = requests.get(url, headers=headers, params=params)
locations = x.json()["stopLocationOrCoordLocation"]
```

```python
for location in locations:
    name = location.get("CoordLocation", {}).get("name", "N/A")
    id = location.get("CoordLocation", {}).get("id", "N/A")
    print(name, id)
```

## Departure

```python
service = "departureBoard"
url = f"https://{domain}/{service}"

params = {    
    "id": "A=2@O=Schusterstraße 55116 Mainz",
    "duration": 30,
    "passlist": 1,
}

x = requests.get(url, headers=headers, params=params)

departures = x.json().get("Departure", [])
```

```python
for dep in departures:
    print(dep.get("stop"))
```

```python
for dep in departures:
    print(dep.get("Messages", {}).get("Message", [{}])[0].get("text"), "N/A")
# departures[0]["Messages"]["Message"][0]["messageCategory"]
```

```python
stps = departures[4].get("Stops").get("Stop", [])
for stp in stps:
    print(stp.get("name"), stp.get("arrTime"), stp.get("id"))
```

```python
stps
```

```python
print(f"{len(departures)} departures found:")
for departure in departures:
    if "Mainz Höfchen" != departure.get("stop"):
        continue
    name = departure.get("name", "N/A")
    direction = departure.get("direction", "N/A")
    scheduled_time = datetime.strptime(departure.get("time", "N/A"), "%H:%M:%S").time() if departure.get("time") else None
    actual_time = datetime.strptime(departure.get("rtTime", "N/A"), "%H:%M:%S").time() if departure.get("rtTime") else None
    if scheduled_time and actual_time:
        delay = (datetime.combine(datetime.today(), actual_time) - datetime.combine(datetime.today(), scheduled_time)).total_seconds() / 60
    else:
        delay = "N/A"
    stop = departure.get("stop", "N/A")
    track = departure.get("track", "N/A")
    # filter duplicates
    # specific stops
    # messages
    print(f"{name} nach {direction}: Scheduled - {scheduled_time}, Actual - {actual_time}, Delay - {delay} min, Stop ID - {stop}, track - {track}")
```
