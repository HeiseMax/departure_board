# RMV API

## Setup

```python
import os
import requests
from datetime import datetime
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("api_key")
```

```python
domain = "www.rmv.de/hapi"

headers = {
    "Authorization": f"Bearer {api_key}",
    "accept": "application/json",
}
```

## Location search

```python
service = "location.name"
url = f"https://{domain}/{service}"

params = {    
    "input": "Schusterstraße",
}

x = requests.get(url, headers=headers, params=params)
locations = x.json()["stopLocationOrCoordLocation"]
```

```python
locations[0]
```

```python
for location in locations:
    name = location.get("CoordLocation", {}).get("name", "N/A")
    id = location.get("CoordLocation", {}).get("id", "N/A")
    print(name, id)
```

## Departure

```python
service = "departureBoard"
url = f"https://{domain}/{service}"

params = {    
    "id": "A=2@O=Höfchen 55116 Mainz",
    "duration": 30,
}

x = requests.get(url, headers=headers, params=params)

departures = x.json().get("Departure", [])
```

```python
departures[0]
```

```python
print(f"{len(departures)} departures found:")
for departure in departures:
    name = departure.get("name", "N/A")
    direction = departure.get("direction", "N/A")
    scheduled_time = datetime.strptime(departure.get("time", "N/A"), "%H:%M:%S").time() if departure.get("time") else None
    actual_time = datetime.strptime(departure.get("rtTime", "N/A"), "%H:%M:%S").time() if departure.get("rtTime") else None
    if scheduled_time and actual_time:
        delay = (datetime.combine(datetime.today(), actual_time) - datetime.combine(datetime.today(), scheduled_time)).total_seconds() / 60
    else:
        delay = "N/A"
    print(f"{name} nach {direction}: Scheduled - {scheduled_time}, Actual - {actual_time}, Delay - {delay} min")
```
